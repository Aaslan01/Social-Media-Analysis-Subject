```python
import pandas as pd

# Read text from file
with open("/content/hw.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()
```

```python
numbers = [int(line.strip()) for line in lines]
```

```python
height_weight_pairs = []
for i in range(0, len(numbers), 2):
    height = numbers[i]/100
    weight = numbers[i+1]/2.2
    height_weight_pairs.append([height, weight])
```

```python
df = pd.DataFrame(height_weight_pairs, columns=["height_m", "weight_kg"])
print(df)
```

```python
# BMI Categories:
# Underweight = <18.5
# Normal weight = 18.5–24.9
# Overweight = 25–29.9
# Obesity = BMI of 30 or greater

Underweight = {}
Normal_weight = {}
Overweight = {}
Obesity = {}
for row in df.itertuples():

    bmi = round(row.weight_kg / (row.height_m ** 2),4)
    df=df.assign(BMI=bmi)
    if bmi < 18.5:
        Underweight[f"Person : {row.Index}"] = bmi
    elif 18.5 <= bmi < 24.9:
        Normal_weight[f"Person : {row.Index}"] = bmi
    elif 24.9 <= bmi < 29.9:
        Overweight[f"Person : {row.Index}"] = bmi
    else:
        Obesity[f"Person : {row.Index}"] = bmi

print(Underweight)
print(Normal_weight)
print(Overweight)
print(Obesity)
```

```python
# Now make function:
def output_BMI(dict_BMI):
  total = 0
  for item, value in dict_BMI.items():
      total += value
      print(f"{item}: {value}")
  print(f"Group Size : {len(dict_BMI)}")
  print(f"Average BMI of this : {round(total/len(dict_BMI),2)}")
```

```python
output_BMI(Underweight)
output_BMI(Normal_weight)
output_BMI(Overweight)
output_BMI(Obesity)
print(f"This file has record for {len(df['BMI'])} persons")
print(f"Average BMI for all these {len(df['BMI'])} persons is", sum(df['BMI'])/len(df['BMI']))
```

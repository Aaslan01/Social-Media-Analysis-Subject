```python
import pandas as pd

df = pd.read_csv("/content/Tweets.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# drop rows where tweet_coord is not empty
df = df.dropna(subset=["tweet_coord"])
```

```python
# rows and columns
df.shape
```

```python
# Loop through the rows and add markers
m = folium.Map(location=[40.74804263, -73.99295302], zoom_start=10)

for index, row in df.iterrows():
    try:
        coords = ast.literal_eval(row["tweet_coord"])  # safely convert string to list
        if len(coords) == 2:
            lat, lon = coords
            folium.Marker(location=[lat, lon],
                          tooltip=row['airline_sentiment'],
                          popup=row['airline'],
                          icon=folium.Icon(color="green"),
                          ).add_to(m)
    except Exception as e:
        continue

# Save map to HTML file
m
```

````python
# ROI Generated

Ad_Spend = 1000
Revenue_Generated = 3500
Likes = 500
Comments = 120
Shares = 80
Reach = 10000
New_Followers = 300
````

````python
# ROI

roi = (Revenue_Generated - Ad_Spend) / Ad_Spend * 100
print(roi,'%')
````

````python
# SMI Cal

engagement_score = 0.5 * Likes + 1.5 * Comments + 2 * Shares
print(engagement_score)
````

````python
# Bar Chart

# Likes, Comments, Shares, Reach, and New Followers
import matplotlib.pyplot as plt
import numpy as np

x_labels = ['Likes', 'Comments', 'Shares', 'Reach', 'New Followers']
y_values = [ Likes, Comments, Shares, Reach, New_Followers]

plt.bar(x_labels, y_values)
plt.xlabel('Metrics')
plt.ylabel('Values')
plt.title('Metrics vs Values')

plt.show()
````
